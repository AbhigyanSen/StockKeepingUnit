#### Matching

```python
import pandas as pd

def analyze_dataframes(ms, master):
    """
    Analyzes two DataFrames to find matching values and calculate metrics.

    Args:
        ms (pd.DataFrame): DataFrame with 'nitemcode' and 'Suggestion' columns.
        master (pd.DataFrame): DataFrame with a single column (name doesn't matter).

    Returns:
        None. Prints the calculated metrics.
    """

    # 1. Find matching values in the same row
    matches = []
    for i in range(len(master)):
        master_value = master.iloc[i, 0]  # Get value from master DataFrame
        if (ms.iloc[i]['nitemcode'] == master_value) or (ms.iloc[i]['Suggestion'] == master_value):
            matches.append(True)
        else:
            matches.append(False)

    # Calculate metrics for all rows
    true_count = matches.count(True)
    false_count = matches.count(False)
    total_count = len(matches)

    true_percentage = (true_count / total_count) * 100 if total_count > 0 else 0
    false_percentage = (false_count / total_count) * 100 if total_count > 0 else 0

    print("Overall Matching Metrics:")
    print(f"True Matches: {true_count} ({true_percentage:.2f}%)")
    print(f"False Matches: {false_count} ({false_percentage:.2f}%)")

    # 2. Analyze rows containing '9'
    nine_matches = []
    for i in range(len(master)):
        master_value = master.iloc[i, 0]
        if master_value == 9:
            if (ms.iloc[i]['nitemcode'] == 9) or (ms.iloc[i]['Suggestion'] == 9):
                nine_matches.append(True)
            else:
                nine_matches.append(False)

    true_nine_count = nine_matches.count(True)
    false_nine_count = nine_matches.count(False)
    total_nine_count = len(nine_matches)

    true_nine_percentage = (true_nine_count / total_nine_count) * 100 if total_nine_count > 0 else 0
    false_nine_percentage = (false_nine_count / total_nine_count) * 100 if total_nine_count > 0 else 0

    print("\nMetrics for Rows Containing '9':")
    print(f"True '9' Matches: {true_nine_count} ({true_nine_percentage:.2f}%)")
    print(f"False '9' Matches: {false_nine_count} ({false_nine_percentage:.2f}%)")

ms = pd.read_csv(r"D:\Projects\SKU\StockKeepingUnit\Result_Verification\Version_2.3.2\matches+suggestion1.csv")  # Replace with your actual file path
master = pd.read_csv(r"D:\Projects\SKU\StockKeepingUnit\Result_Verification\Version_2.3.2\master1.csv")  # Replace with your actual file path

# Crop master DataFrame
ms_len = len(ms)
master = master.iloc[:ms_len]

analyze_dataframes(ms, master)
```

### Surfing the **Complete Results**

```python
import pandas as pd

def analyze_dataframes(ms, master):
    """
    Analyzes two DataFrames to find matching values and calculate metrics.
    Concatenates the DataFrames and adds a flag column for true and false matches.

    Args:
        ms (pd.DataFrame): DataFrame with 'nitemcode' and 'Suggestion' columns.
        master (pd.DataFrame): DataFrame with a single column (name doesn't matter).

    Returns:
        pd.DataFrame: Concatenated DataFrame with a flag column indicating matches.
    """

    # 1. Find matching values in the same row
    matches = []
    flags = []
    for i in range(len(master)):
        master_value = master.iloc[i, 0]  # Get value from master DataFrame
        if (ms.iloc[i]['nitemcode'] == master_value) or (ms.iloc[i]['Suggestion'] == master_value):
            matches.append(True)
            flags.append(0)  # True match flag
        else:
            matches.append(False)
            flags.append(1)  # False match flag

    # Calculate metrics for all rows
    true_count = matches.count(True)
    false_count = matches.count(False)
    total_count = len(matches)

    true_percentage = (true_count / total_count) * 100 if total_count > 0 else 0
    false_percentage = (false_count / total_count) * 100 if total_count > 0 else 0

    print("Overall Matching Metrics:")
    print(f"True Matches: {true_count} ({true_percentage:.2f}%)")
    print(f"False Matches: {false_count} ({false_percentage:.2f}%)")

    # 2. Concatenate DataFrames and add flag column
    concatenated_df = pd.concat([ms, master.reset_index(drop=True)], axis=1)
    concatenated_df['Match_Flag'] = flags

    return concatenated_df

# Assuming ms and master DataFrames are loaded from CSVs as in your original code
ms = pd.read_csv(r"D:\Projects\SKU\StockKeepingUnit\Result_Verification\Version_2.3.2\matches+suggestion2.csv")
master = pd.read_csv(r"D:\Projects\SKU\StockKeepingUnit\Result_Verification\Version_2.3.2\master2.csv")

# Crop master DataFrame
ms_len = len(ms)
master = master.iloc[:ms_len]

# Analyze and concatenate DataFrames
result_df = analyze_dataframes(ms, master)
result_df.to_csv(r"D:\Projects\SKU\StockKeepingUnit\Result_Verification\Version_2.3.2\concatenated_results.csv", index=False)
```
